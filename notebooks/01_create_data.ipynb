```python
# ========================================
# STEP 1: GENERATE SMARTPHONE DATA (FIXED)
# ========================================

import pandas as pd
import numpy as np
import random

# Set seed for same results every time
np.random.seed(42)
random.seed(42)

# Define categories
brands = ['Apple', 'Samsung', 'OnePlus', 'Google', 'Xiaomi']
conditions = ['Excellent', 'Good', 'Fair', 'Poor']

print("🔄 Generating smartphone data...")

# Create 1500 records
data = []
for i in range(1500):
    brand = random.choice(brands)
    age = random.randint(0, 48)
    ram = random.choice([4, 6, 8, 12])
    storage = random.choice([64, 128, 256])
    battery = random.randint(70, 100)
    
    # FIXED THIS LINE - Added 's' to choice and used weights
    condition = random.choices(conditions, weights=[0.3, 0.4, 0.2, 0.1])[0]
    
    original_price = random.randint(20000, 100000)
    
    # Calculate resale price
    depreciation = 1 - (age / 48) * 0.6
    cond_mult = {'Excellent': 1.0, 'Good': 0.85, 'Fair': 0.7, 'Poor': 0.5}
    resale = original_price * depreciation * cond_mult[condition]
    resale = resale + random.randint(-2000, 2000)
    resale = max(resale, 1000)
    
    data.append([brand, age, ram, storage, battery, condition, original_price, int(resale)])

# Create DataFrame
columns = ['brand', 'age_months', 'ram_gb', 'storage_gb', 'battery_health', 
           'condition', 'original_price', 'resale_price']
df = pd.DataFrame(data, columns=columns)

# Save to CSV
df.to_csv('../data/smartphone_data.csv', index=False)

print("✅ Data created successfully!")
print(f"📊 Total records: {len(df)}")
print("\n📋 First 5 rows:")
print(df.head())
print("\n📈 Summary:")
print(df.describe())
```
