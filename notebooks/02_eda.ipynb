```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load the data
df = pd.read_csv('../data/smartphone_data.csv')

# Create plots folder (if not exists)
os.makedirs('../app/static/plots', exist_ok=True)

print("✅ Data loaded successfully!")
print(f"📊 Shape: {df.shape}")
print("\n📋 First 5 rows:")
df.head()
```

```python
# Plot 1: Price Distribution
plt.figure(figsize=(10, 6))
sns.histplot(df['resale_price'], bins=30, kde=True, color='blue')
plt.title('📊 Smartphone Resale Price Distribution', fontsize=14)
plt.xlabel('Resale Price (₹)', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.grid(True, alpha=0.3)

# Add mean and median lines
mean_price = df['resale_price'].mean()
median_price = df['resale_price'].median()
plt.axvline(mean_price, color='red', linestyle='--', label=f'Mean: ₹{mean_price:,.0f}')
plt.axvline(median_price, color='green', linestyle='--', label=f'Median: ₹{median_price:,.0f}')
plt.legend()

plt.savefig('../app/static/plots/price_distribution.png', dpi=100, bbox_inches='tight')
plt.show()
print("✅ Plot 1 saved: price_distribution.png")
```

```python
# Plot 2: Brand vs Price
plt.figure(figsize=(10, 6))
sns.boxplot(x='brand', y='resale_price', data=df, palette='Set2')
plt.title('📊 Resale Price by Brand', fontsize=14)
plt.xlabel('Brand', fontsize=12)
plt.ylabel('Resale Price (₹)', fontsize=12)
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

# Add average price on top
avg_by_brand = df.groupby('brand')['resale_price'].mean().sort_values()
print("\n💰 Average Price by Brand:")
print(avg_by_brand)

plt.savefig('../app/static/plots/brand_price.png', dpi=100, bbox_inches='tight')
plt.show()
print("✅ Plot 2 saved: brand_price.png")
```

```python
# Plot 3: Age vs Price (with condition colors)
plt.figure(figsize=(10, 6))
scatter = sns.scatterplot(x='age_months', y='resale_price', 
                         hue='condition', data=df, alpha=0.6, s=50)
plt.title('📊 Phone Age vs Resale Price (by Condition)', fontsize=14)
plt.xlabel('Age (months)', fontsize=12)
plt.ylabel('Resale Price (₹)', fontsize=12)
plt.legend(title='Condition')
plt.grid(True, alpha=0.3)

plt.savefig('../app/static/plots/age_price.png', dpi=100, bbox_inches='tight')
plt.show()
print("✅ Plot 3 saved: age_price.png")
```

```python
# Plot 4: Condition vs Price
plt.figure(figsize=(10, 6))
order = ['Excellent', 'Good', 'Fair', 'Poor']
sns.boxplot(x='condition', y='resale_price', data=df, order=order, palette='Set3')
plt.title('📊 Resale Price by Phone Condition', fontsize=14)
plt.xlabel('Condition', fontsize=12)
plt.ylabel('Resale Price (₹)', fontsize=12)
plt.grid(True, alpha=0.3)

# Add average price by condition
print("\n💰 Average Price by Condition:")
print(df.groupby('condition')['resale_price'].mean().sort_values(ascending=False))

plt.savefig('../app/static/plots/condition_price.png', dpi=100, bbox_inches='tight')
plt.show()
print("✅ Plot 4 saved: condition_price.png")
```

```python
# Plot 5: Correlation Heatmap
plt.figure(figsize=(10, 8))
numeric_cols = ['age_months', 'ram_gb', 'storage_gb', 'battery_health', 
                'original_price', 'resale_price']
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', square=True, 
            linewidths=0.5, cbar_kws={"shrink": 0.8})
plt.title('📊 Feature Correlation Heatmap', fontsize=14)

plt.savefig('../app/static/plots/correlation.png', dpi=100, bbox_inches='tight')
plt.show()
print("✅ Plot 5 saved: correlation.png")
```

```python
# Bonus: RAM vs Price
plt.figure(figsize=(10, 6))
sns.boxplot(x='ram_gb', y='resale_price', data=df, palette='coolwarm')
plt.title('📊 Resale Price by RAM Size', fontsize=14)
plt.xlabel('RAM (GB)', fontsize=12)
plt.ylabel('Resale Price (₹)', fontsize=12)
plt.grid(True, alpha=0.3)

# Add average price by RAM
print("\n💰 Average Price by RAM:")
print(df.groupby('ram_gb')['resale_price'].mean())

plt.savefig('../app/static/plots/ram_price.png', dpi=100, bbox_inches='tight')
plt.show()
print("✅ Bonus plot saved: ram_price.png")
```

```python
# Bonus 2: Storage vs Price
plt.figure(figsize=(10, 6))
sns.boxplot(x='storage_gb', y='resale_price', data=df, palette='viridis')
plt.title('📊 Resale Price by Storage Size', fontsize=14)
plt.xlabel('Storage (GB)', fontsize=12)
plt.ylabel('Resale Price (₹)', fontsize=12)
plt.grid(True, alpha=0.3)

print("\n💰 Average Price by Storage:")
print(df.groupby('storage_gb')['resale_price'].mean())

plt.savefig('../app/static/plots/storage_price.png', dpi=100, bbox_inches='tight')
plt.show()
print("✅ Bonus 2 plot saved: storage_price.png")
```
